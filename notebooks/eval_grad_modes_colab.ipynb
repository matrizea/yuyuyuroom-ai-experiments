```python
import hashlib, json, platform, urllib.request
import torch

url = "https://raw.githubusercontent.com/matrizea/yuyuyuroom-ai-experiments/4529519e799cef59cba1962ac37ea2123a941625/ai_lab/experiment_eval_grad_modes.py"
source = urllib.request.urlopen(url, timeout=30).read()
assert hashlib.sha256(source).hexdigest() == "904d4f9461eca6aee2c9855faf20d9e0631a0629aa66d6bf6608f3871e7e2fe6"
probe = {"__name__": "eval_probe"}
exec(compile(source.decode("utf-8"), "experiment_eval_grad_modes.py", "exec"), probe)
result = {"environment": {"python": platform.python_version(), "torch": torch.__version__, "cuda_available": torch.cuda.is_available(), "gpu": torch.cuda.get_device_name(0) if torch.cuda.is_available() else None}, "devices": {}}
for name in (["cpu", "cuda"] if torch.cuda.is_available() else ["cpu"]):
    device = torch.device(name)
    result["devices"][name] = {
        "mode_combinations": probe["run_six_mode_combinations"](device),
        "tensor_reuse_cases": probe["run_tensor_reuse_subexperiment"](device),
    }
print(json.dumps(result, ensure_ascii=False, indent=2))
```
